```python
import re
import numpy as np
from PIL import Image

def clean_plate_text(text):
    text = text.upper()
    text = re.sub(r'[^A-Z0-9]', '', text)
    return text


def detect_and_read_plate(image_path):

    # YOLO11 detects license plates
    results = best_model.predict(
        source=image_path,
        conf=0.25,
        verbose=False
    )

    result = results[0]

    # Open original image
    image = Image.open(image_path).convert("RGB")

    detected_plates = []

    # Loop through detected plates
    for box in result.boxes:

        # Get coordinates
        x1, y1, x2, y2 = (
            box.xyxy[0]
            .cpu()
            .numpy()
            .astype(int)
        )

        # Crop license plate
        plate_crop = image.crop(
            (x1, y1, x2, y2)
        )

        # Convert PIL Image → NumPy array
        plate_crop_np = np.array(plate_crop)

        # EasyOCR
        ocr_results = reader.readtext(
            plate_crop_np
        )

        # Process OCR results
        for bbox, text, confidence in ocr_results:

            cleaned_text = clean_plate_text(text)

            if cleaned_text:
                detected_plates.append({
                    "plate": cleaned_text,
                    "ocr_confidence": float(confidence),
                    "bbox": [x1, y1, x2, y2]
                })

    return detected_plates
```

```python
plates = detect_and_read_plate(image_name)

print(plates)
```
